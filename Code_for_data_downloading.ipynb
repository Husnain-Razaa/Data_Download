```python
import os
import re
import nltk
import pandas as pd

# Download NLTK resources (word_tokenize and punkt tokenizer)
nltk.download('punkt')

# Function to extract information from each text file
def extract_info_from_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()

    # Extract Title, Date, and File Name
    title_match = re.search(r"Title:(.+)", content, re.IGNORECASE)
    title = title_match.group(1).strip() if title_match else ""

    date_match = re.search(r"Published in (.+), (.+ \d{1,2}(?:st|nd|rd|th), \d{4})", content, re.IGNORECASE)
    date = date_match.group(2) if date_match else ""

    file_name = os.path.basename(file_path)

    # Extract Writer Occupation and Writer Other Info
    writer_occupation = ""
    writer_other_info = ""
    occupation_index = content.find("The writer")
    if occupation_index != -1:
        writer_occupation_match = re.search(r"The writer (is|teaches|works as) (.+?)[.]", content[occupation_index:], re.IGNORECASE)
        if writer_occupation_match:
            writer_occupation = writer_occupation_match.group(2).strip()
            writer_other_info = content[occupation_index + writer_occupation_match.end():].strip()

    # Extract Word Count
    word_count = len(nltk.word_tokenize(content.split("Title:", 1)[0]))

    return {
        "Title": title,
        "Date": date,
        "File Name": file_name,
        "Writer_Occupation": writer_occupation,
        "Writer_other_info": writer_other_info,
        "Word Count": word_count
    }

# Path to the folder containing text files
folder_path = "/content/drive/MyDrive/The News"

# List to store information from each file
data = []

# Process each file in the folder
for file_name in os.listdir(folder_path):
    if file_name.endswith(".txt"):
        file_path = os.path.join(folder_path, file_name)
        file_info = extract_info_from_file(file_path)
        data.append(file_info)

# Convert the data to a Pandas DataFrame
df = pd.DataFrame(data)

# Save the DataFrame to an Excel file
output_file = "The News Meta Data.xlsx"
df.to_excel(output_file, index=False)
```

```python
def count_words_in_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
        word_count = len(nltk.word_tokenize(content))
    return word_count

# Path to the folder containing .txt files
folder_path = "/content/drive/MyDrive/The News"

# List to store file names and word counts
data = []

# Process each file in the folder
for file_name in os.listdir(folder_path):
    if file_name.endswith(".txt"):
        file_path = os.path.join(folder_path, file_name)
        word_count = count_words_in_file(file_path)
        data.append({"File": file_name, "Word Count": word_count})

# Create a Pandas DataFrame
df = pd.DataFrame(data)

# Save the DataFrame to an Excel file
output_file = "word_count_output_the_news.xlsx"
df.to_excel(output_file, index=False)

print("Word count information saved to word_count_output.xlsx")
```
